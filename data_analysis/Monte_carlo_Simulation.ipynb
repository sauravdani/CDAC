```python
import numpy as np
```

```python
def roll_dice():
    #Simulatipn of rolling a dice twice, minimum noumber we get is 1, max is 6 
    #add the result of the two sumulations, so possible values we be:
    #(1,1) or (1,2) ... 
    return np.sum(np.random.randint(1,7,2))

print(roll_dice())
```

```python

#making monte carlo simulation for the game to get number 7 to get $5 else loosing $1
def monte_carlo_simulation(runs = 1000):
    results = np.zeros(2)
    for _ in range(runs):
        if (roll_dice() == 7):
            results[0] += 1
        else:
            results[1] += 1
    return results
```

```python
print(monte_carlo_simulation())
print(monte_carlo_simulation())
print(monte_carlo_simulation())
```

```python
result = np.zeros(1000)
for i in range(1000):
    result[i] = monte_carlo_simulation()[0]

print(result)
```

```python
#let us plot it
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.hist(result, bins=15)
plt.show()
```

```python
print(result.mean())
print(result.mean() * 5)
print(1000- result.mean())
print(result.mean()*4.75)
print(1000-result.mean())
print(result.mean()/1000)
```

```python
#Fincancial investments modeling using Monte Carlo
import numpy as np
import matplotlib.pyplot as plt
```

```python
investment = 100000     #100000 investing
num_years = 20
stock_weight = 0.6

bond_weight = 0.4
```

```python
stock_retun = 0.5
stock_sd = 0.60
stock_weight = 0.07
bond_return = 0.07
bond_sd = 0.05
```

```python
iteration = 5555
```

```python
stock_retun = np.random.normal(stock_retun, stock_sd, (iteration, num_years))
bond_return = np.random.normal(bond_return, bond_sd, (iteration, num_years))

print(stock_retun)
print(bond_return)
total_return_pct = ((stock_weight * stock_retun) + (bond_weight * bond_return)) * 100
```

```python
#suppress scinetific
np.set_printoptions(suppress=True)
```

```python
#Create a 5555 * 20 array that will sotre cumulative investement values
portfolio_values = investment * (1 + (total_return_pct/100)).cumprod(axis=1)
print("***portfolio Vaules")
print(portfolio_values)
```

```python
final_values = portfolio_values[:, -1]
print("Summary of final postfolioio values after {num_years} years: ")
print("Mean: ", final_values.mean())
print("Standard Deviation: ", final_values)
```

```python
#Calculate percentiles (eg. 25th, 50th and 100th percentiles)
percentiles = np.percentile(final_values, [25, 50, 100])
print("percentiles off final portfolios values: ")
print("25th percentile: ", percentiles[0])
print("50th percentile: ", percentiles[1])
print("100th percentile: ", percentiles[2])
```

#Startup Investment Decision: Monte Carlo

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Parameters
inetia_investment = 1000000
success_rate = 0.01
failure_loss = 0.5
success_multiplier = 1000000
num_simulations = 10000
```

```python
#Monte carlo simulation
final_values = []
for _ in range(num_simulations):
    success = np.random.choice([1,0], p=[success_rate, 1-success_rate])
    if(success):
        final_value = inetia_investment * success_multiplier
    else:
        final_value = inetia_investment * (1-failure_loss)
    final_values.append(final_value)
```

```python
#Results
mean_final_value = np.mean(final_values)
risk_of_loss = np.mean(np.array(final_value) < inetia_investment)

print("Mean Final Value: ", mean_final_value)
print("Probability of loss: ", risk_of_loss * 100, "%")
```

```python
#Plotting the destribution
plt.hist(final_values, bins=30, edgecolor = 'k')
plt.title("Simulated startup investment Outcomes")
plt.xlabel("Investment Vaues; ")
plt.ylabel("Frequency")
plt.show()
```
