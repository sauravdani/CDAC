one way ANOVA TEST 
for ANOVA there is no need to use the sample -use the full dataset

```python
from scipy.stats import f_oneway
import pandas as pd
```

```python
df=pd.read_csv("cust_seg.csv")
```

```python
print(df.segment.value_counts())
s1=df.Latest_mon_usage[df.segment==1]
s2=df.Latest_mon_usage[df.segment==2]
s3=df.Latest_mon_usage[df.segment==3]
```

```python
fstat,pval=f_oneway(s1,s2,s3)
alpha=0.05
if pval<alpha:
    print("Reject HO")
else:
    print("DO not reject HO")
```
