```python
# Subtract row mean from each row of a 2D array
array = np.arange(1,17)
array_2d = array.reshape(4,4)
print(array_2d)
print(np.mean(array_2d, axis=1))

array_2d = array_2d[:,:] - np.mean(array_2d, axis=1)
array_2d
```

```python
#Given a 4×4 array, swap the first half rows with the second half rows
array = np.arange(1,17)
array_2d = array.reshape(4,4)
first = array_2d[:2, :]
second = array_2d[2:,:]
print (first)
print (second)
ans = np.vstack([second, first])
ans
```
