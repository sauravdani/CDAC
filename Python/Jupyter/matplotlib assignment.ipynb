```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('Population.csv')
```

```python
df
```

```python
# Draw line & bar charts to show urban and rural population of India over the years using population.csv 
#[rural population is the difference between total population and urban population][For bar chart consider 8 yrs data]
df['r_pop']=df['Population']-df['Urban Population']
plt.plot(df['Year'], df['Urban Population'], label="urban")
plt.plot(df['Year'], df['r_pop'], label="Rural")
plt.title("Graph for urban and rural population")
plt.xlabel("Years")
plt.ylabel("Population in Millions")
plt.legend()
plt.show()
```

```python
df["r_pop"]=df["Population"]-df['Urban Population']
plt.bar(df['Year'],df['Urban Population'],label="Urban")

plt.bar(df['Year'], df['r_pop'], bottom=df['Urban Population'], label="Rural")
plt.xlabel("Year")
plt.ylabel("population in Million")
plt.legend()
plt.title("Graph for urban and rural population")
plt.show
```

```python
#Using gapminder.csv show life expectancy of top 10 highly populated countries from Asia
df = pd.read_csv('gapminder.csv')

asian_10=df[df['continent']=='Asia']
asian_10 = asian_10.sort_values(by='lifeExp', ascending = False)
asian_10
asian_10 = asian_10.head(10)
asian_10
```

```python
plt.bar(asian_10["country"], asian_10["lifeExp"], color="#9b47f5")
plt.xticks(rotation = 60)
plt.xlabel("Country Names")
plt.ylabel("Life Expetency")
plt.title("Lige Expetency per country")

plt.show
```
