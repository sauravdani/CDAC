Q1. Create a DataFrame containing student information like Name, Age, Department, %Marks
Display column names of the DF
Show min/max and average marks
Get unique values and student count for each Department


```python
data=[{'name':"Max", "Age":22, "dept":"DBDA","Marks":93},
    {'name':"Sam", "Age":19, "dept":"DAC","Marks":63},
      {'name':"Bob", "Age":24, "dept":"DITIS","Marks":73},
      {'name':"NOVA", "Age":26, "dept":"VLSI","Marks":82},
     ]

df = pd.DataFrame(data)
df
```

```python
df.columns
```

```python
min_marks = df["Marks"].min()
print("max marks: ",min_marks)

max_marks = df["Marks"].max()
print("max marks: ",max_marks)

avg_marks = df["Marks"].mean() #mean is use for find avarage value
print("avg marks: ",avg_marks)
```

Q2. Read csv file solar.csv
Display all the plants with capacity > 500 MW
Display plant details for New York plant
For all the plants display Average MW Per Plant and Generation (GWh)
Sort the plants by Generation (GWh) (ascending) and Installed Capacity (MW)(descending)
List top 5 plants for their Generation
Display all plant details for states – California, Nevada, Arizona and Texas
Group the plants by region, find – min/max capacity plant for each group
Add following information to the existing details for the plants


```python
df=pd.read_csv('solar.csv')
df
```

```python
#Display all the plants with capacity > 500 MW
data=df[df['Installed Capacity (MW)']>500]
data
```

```python
#Display plant details for New York plant
newyork_plant=df[df['State']=='New York']
newyork_plant
```

```python
#For all the plants display Average MW Per Plant and Generation (GWh)
data1=df[['State','Average MW Per Plant','Generation (GWh)'], set_index='State']
data1
```

```python
#Sort the plants by Generation (GWh) (ascending) and Installed Capacity (MW)(descending)
sorted=df.sort_values(by=['Generation (GWh)','Installed Capacity (MW)'], ascending=[True,False])
sorted
```

```python
#List top 5 plants for their Generation
top5=df.sort_values(by='Generation (GWh)', ascending=False).head(5)
top5
```

```python
#Display all plant details for states – California, Nevada, Arizona and Texas
data2=df[df['State'].isin( ['California','Nevada','Arizona','Texas'])]
data2
```

```python
#Group the plants by region, find – min/max capacity plant for each group
data3 = df.groupby('Region')['Installed Capacity (MW)'].min()
print(data3)

data3 = df.groupby('Region')['Installed Capacity (MW)'].max()
print(data3)
```
