```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# 1. Load the data
df = pd.read_csv('Population.csv')

# 2. Calculate Rural Population
df['Rural Population'] = df['Population'] - df['Urban Population']

# Select the necessary columns for plotting
df_plot = df[['Year', 'Urban Population', 'Rural Population']]

# Define a formatter function for large population numbers (Millions/Billions)
def millions_formatter(x, pos):
    if x >= 1e9:
        # Convert to Billions
        return f'{x/1e9:.1f}B'
    elif x >= 1e6:
        # Convert to Millions
        return f'{x/1e6:.0f}M'
    return f'{x:.0f}'

# --- Line Chart (All Years) ---
plt.figure(figsize=(10, 6))

# Plot both urban and rural populations
plt.plot(df_plot['Year'], df_plot['Urban Population'], label='Urban Population', marker='o')
plt.plot(df_plot['Year'], df_plot['Rural Population'], label='Rural Population', marker='o')

# Add titles and labels
plt.title('India: Urban vs. Rural Population Over the Years (Line Chart)')
plt.xlabel('Year')
plt.ylabel('Population')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# Apply the custom formatter to the y-axis
plt.gca().yaxis.set_major_formatter(FuncFormatter(millions_formatter))

# Set x-ticks to show every 5th year for better readability
plt.xticks(df_plot['Year'][::5]) 
plt.tight_layout()
plt.savefig('urban_rural_line_chart.png')
plt.close()

# --- Bar Chart (Last 8 Years) ---
# Select the last 8 years of data as requested
df_bar = df_plot.tail(8).reset_index(drop=True)

plt.figure(figsize=(12, 7))
bar_width = 0.35
r1 = range(len(df_bar))
# Calculate positions for the second set of bars
r2 = [x + bar_width for x in r1]

# Plot the Urban Population bars
plt.bar(r1, df_bar['Urban Population'], color='skyblue', width=bar_width, edgecolor='grey', label='Urban Population')
# Plot the Rural Population bars
plt.bar(r2, df_bar['Rural Population'], color='sandybrown', width=bar_width, edgecolor='grey', label='Rural Population')

# Add titles and labels
plt.title(f'India: Urban vs. Rural Population ({df_bar["Year"].min()} - {df_bar["Year"].max()}) (Bar Chart)')
plt.xlabel('Year', fontweight='bold')
plt.ylabel('Population', fontweight='bold')

# Set x-axis labels to the year in the center of the grouped bars
plt.xticks([r + bar_width/2 for r in range(len(df_bar))], df_bar['Year'])
plt.legend()
plt.gca().yaxis.set_major_formatter(FuncFormatter(millions_formatter)) # Apply the formatter
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('urban_rural_bar_chart.png')
plt.close()

# Note: The original code saved the files to disk.
```
