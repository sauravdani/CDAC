# For two variables

```python
x = [0.2, 1.2,1,1.4,-1.5,0.5,-0.5]
y = [i*3 + 5 for i in x]

# original_df = pd.DataFrame(data = x, columns=['x'])
# original_df['y'] = y
# print(original_df)
```

```python
eta = 0.2
b = 0.1
w = 0.3
```

Batch Gradient Descent

```python
for i in range(100):
    
    y_pred = [_*w + b for _ in x]
    error = []
    error_x = []

    for _ in range(len(y)):
        error.append(y[_] - y_pred[_])
        error_x.append(error[_] * x[_])


    error_sq = [i*i for i in error]

    db = -np.mean(error)
    dw = -np.mean(error_x)
    l = np.mean(error_sq)/2

    new_b = b - eta * db
    new_w = w - eta * dw

    # print(f'l = {l}, db = {db}, dw = {dw}, current_b = {new_b}, current_w = {new_w}')

    if i % 10 == 0:  # Print every 10 iterations
        print(f'Iteration {i+1}: loss={l:.4f}, b={b:.4f}, w={w:.4f}')

    b = new_b
    w = new_w

print(f'Iteration {i+1}: loss={l:.4f}, b={b:.4f}, w={w:.4f}')
```

# In case of 2 features

```python
x1 = np.linspace(-0.6, 0.7, 10)
x2 = np.linspace(-0.9,0.92, 10)
eta = 0.2
w1 = 0.3
w2 = 0.1
y = 5 + 8 * x1 + 9 * x2
# print(y)

for i in range(100):
    y_pred = b + w1 * x1 + w2 * x2
    L = np.mean((y - y_pred)**2)/2
    dw1 = -np.mean((y-y_pred)*x1)
    dw2 = -np.mean((y-y_pred)*x2)
    db = -np.mean(y - y_pred)

    print(f"{i+1}th iteration :")
    print(f"Loss = {L}, dw1 = {dw1}, dw2 = {dw2}, db = {db}")
    print(f"w1 = {w1}, w2 = {w2}, b = {b}")

    if L < 0.0001:
        break

    new_w1 = w1 - eta*dw1
    new_w2 = w2 - eta*dw2
    new_b = b - eta*db

    w1 = new_w1
    w2 = new_w2
    b = new_b
```

### Stochastic Gradient Descent

```python
# for single variable only

x = np.array([0.2, 1.2,1,1.4,-1.5,0.5,-0.5])
y = 3 * x + 5
eta = 0.2
w,b = 0.3, 0.1
tol = 0.0001

tol = 0.0001
n = len(x)

for i in range(0,50):
    for j in range(0,len(x)):
        y_pred = w* x[j] + b
        err = y[j] - y_pred
        dw, db = -(err*x[j]),-err
        new_w , new_b= w - eta*dw, b - eta*db
        w,b = new_w, new_b
    y_pred_all = w*x + b
    err_all = y - y_pred_all
    L = np.sum(err_all**2)/(2*n)
    print("w = ", w, 'b = ', b)
    print('Loss = ', L)
    if L < tol:
        print(f"Algo stopped at {i+1}th iteration as Loss < Tolerance {tol}")
        break
```

Mini Batch Gradient Descent

```python
x = np.array([0.2,1.2,1,1.4,-1.5,0.5,-0.5])
y = 3*x + 5

w = 0.3
b = 0.1
batches = [[0,6], [3,1,5],[4,2]]
tol = 0.0001

for i in range(0,50):
    for j in range(0, len(x)):
        for batch in batches:
            y_pred = w*x[batch] +b
            err = y[batch] - y_pred
            dw, db = -np.sum(err*x[batch])/len(batch), -np.sum(err)/len(batch)
            new_w, new_b = w - eta* dw, b - eta*db
            w,b = new_w, new_b
    y_pred_all = w*x + b
    err_all = y-y_pred_all
    L = np.sum(err_all**2)/(2*n)
    print("w = ", w, ", b=", b)
    print("Loss = ", L)
    if L < tol:
        print(f'Algo stopped at {i+1}th iteration as Loss < Tolerance {tol}')
        break
    
```
