```python
def gd(x, y, w, b,eta=0.2, iter=100):
    all = []
    iter_num=0
    for i in range(iter):
        iter_num +=1
        y_pred = w*x+b
        err = y-y_pred
        err_sq = err**2
        err_x = err*x
        loss = np.mean(err_sq)/2
        dw = -np.mean(err_x)
        db = -np.mean(err)
        new_w = w - eta*dw
        new_b = b - eta*db
        w = new_w
        b = new_b

        all.append([iter_num, w, b, loss, dw, db, new_w, new_b])
        if loss < 0.0001:
            break
    return pd.DataFrame(all, columns=['Iteration Num', 'w', 'b','Loss', 'dw', 'db', 'new_w', 'new_b'])

gd(x,y,w,b)
```

```python
x1 = np.linspace(-0.6, 0.7, 10)
x2 = np.linspace(-0.9, 0.92, 10)
y = 5 + 8*x1 + 9*x2

eta = 0.2
w1, w2, b = 0.3, 0.1, -0.3
```

```python
def gd(x1, x2, y, w1, w2, b,eta=0.2, iter=100):
    all = []
    iter_num=0
    for i in range(iter):
        iter_num +=1
        y_pred = w1*x1 + w2*x2 + b
        err = y-y_pred
        err_sq = err**2
        err_x1 = err*x1
        err_x2 = err*x2
        loss = np.mean(err_sq)/2
        dw1 = -np.mean(err_x1)
        dw2 = -np.mean(err_x2)
        db = -np.mean(err)
        new_w1 = w1 - eta*dw1
        new_w2 = w2 - eta*dw2
        new_b = b - eta*db
        w1 = new_w1
        w2 = new_w2
        b = new_b
        all.append([iter_num, w1, w2, b, loss, dw1, dw2, db, new_w1, new_w2, new_b])
        if loss < 0.0001:
            break
    return pd.DataFrame(all, columns=['Iteration Num', 'w1', 'w2', 'b','Loss', 'dw1','dw2', 'db', 'new_w1','new_w2', 'new_b'])

gd(x1,x2,y,w1, w2,b)
```

Stochastic Gradient Descent

```python
x = np.array([0.2,1.2,1,1.4,-1.5,0.5,-0.5])
y = 3*x + 5

w = 0.3
b = 0.1
eta = 0.2
tol = 0.0001
n = len(x)
```

```python
for i in range(0,50):
    for j in range(0, len(x)):
        y_pred = w*x[j] +b
        err = y[j] - y_pred
        dw, db = -(err*x[j]), -(err)
        new_w, new_b = w - eta* dw, b - eta*db
        w,b = new_w, new_b
    y_pred_all = w*x + b
    err_all = y-y_pred_all
    L = np.sum(err_all**2)/(2*n)
    print("w = ", w, ", b=", b)
    print("Loss = ", L)
    if L < tol:
        print(f'Algo stopped at {i+1}th iteration as Loss < Tolerance {tol}')
        break
```

Mini-Batch GD

```python
x = np.array([0.2,1.2,1,1.4,-1.5,0.5,-0.5])
y = 3*x + 5

w = 0.3
b = 0.1
batches = [[0,6], [3,1,5],[4,2]]
tol = 0.0001
```

```python
for i in range(0,50):
    for j in range(0, len(x)):
        for batch in batches:
            y_pred = w*x[batch] +b
            err = y[batch] - y_pred
            dw, db = -np.sum(err*x[batch])/len(batch), -np.sum(err)/len(batch)
            new_w, new_b = w - eta* dw, b - eta*db
            w,b = new_w, new_b
    y_pred_all = w*x + b
    err_all = y-y_pred_all
    L = np.sum(err_all**2)/(2*n)
    print("w = ", w, ", b=", b)
    print("Loss = ", L)
    if L < tol:
        print(f'Algo stopped at {i+1}th iteration as Loss < Tolerance {tol}')
        break
    
```
