```python
import pandas as pd
import numpy as np
```

```python
def f(x):
    return np.exp(x) - np.log(x)

def fd(x):  #f'(x)
    return np.exp(x) - (1/x)

x0 = 1.5
x_arr = []

def x_next(x):
    return x - 0.01 * fd(x)

cnt = 1
x = x0
while cnt <= 50:
    x_arr.append([x,f(x)])
    x = x_next(x)
    cnt += 1

for i in x_arr:
    print(f'x = {i[0]}, f(x) = {i[1]}')
# print(x_arr)
```
