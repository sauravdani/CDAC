#### Boston
1. train-test_split the dataset
2. Test MAE for degrees = 2,3,4

here we checked wioith different caues of degree to check which gives lowest MAE

```python
boston = pd.read_csv("Datasets/Boston.csv")
boston
```

```python
X = boston.drop('medv', axis=1)
y = boston["medv"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)
```

```python
poly = PolynomialFeatures(degree=2, include_bias=False).set_output(transform='pandas')
X_poly_train = poly.fit_transform(X_train)
X_poly_test = poly.transform(X_test)
lr = LinearRegression()
lr.fit(X_poly_train, y_train)
y_pred = lr.predict(X_poly_test)
mean_absolute_error(y_test, y_pred)
```

#### Housing

```python
housing = pd.read_csv('Datasets/Housing.csv')
housing
```

As here we have different coloumns that have categorical values
So we use the One-Hot encoding

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
#this should not be done as It is nto making one hot encoding only for categorical coloumns but also for the differet numerical coloumns and making 521 coloumns after encoding 
#for every price

ohe = OneHotEncoder(drop='first', sparse_output=False).set_output(transform='pandas')
ohe.fit_transform(housing)
```

```python
#checking the coloumns having only object values
cat_hous = housing[housing.columns[housing.dtypes == object]]
num_hous = housing[housing.columns[housing.dtypes != object]]
```

```python

ohe = OneHotEncoder(drop='first', sparse_output=False).set_output(transform='pandas')
ohe_hous = ohe.fit_transform(cat_hous)
hous = pd.concat([ohe_hous, num_hous], axis=1)
X, y = hous.drop('price', axis=1), hous['price']
hous.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)
```

```python
poly = PolynomialFeatures(degree=1, include_bias=False).set_output(transform='pandas')
X_poly_train = poly.fit_transform(X_train)
X_poly_test = poly.transform(X_test)
lr = LinearRegression()
lr.fit(X_poly_train, y_train)
y_pred = lr.predict(X_poly_test)
mean_absolute_error(y_test, y_pred)
```

```python
#performing best way to work with categorical data
ohe = OneHotEncoder(drop='first', sparse_output=False).set_output(transform='pandas')
col_trnf = ColumnTransformer([('OHE', ohe, list(housing.columns[housing.dtypes == object]))],
                             remainder = 'passthrough',
                             verbose_feature_names_out=False)
col_trnf = col_trnf.set_output(transform='pandas')
hous = col_trnf.fit_transform(housing)
```

```python
X, y = hous.drop('price', axis=1), hous['price']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)
```

```python
poly = PolynomialFeatures(degree=1, include_bias=False).set_output(transform='pandas')
X_poly_train = poly.fit_transform(X_train)
X_poly_test = poly.transform(X_test)
lr = LinearRegression()
lr.fit(X_poly_train, y_train)
y_pred = lr.predict(X_poly_test)
mean_absolute_error(y_test, y_pred)
```
