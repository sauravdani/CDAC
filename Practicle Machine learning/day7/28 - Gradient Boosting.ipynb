```python
glass = pd.read_csv('C:/Users/dbda.STUDENTSDC/Desktop/M.L/Cases/Glass_Identification/Glass.csv')
glass
```

```python
x,y = glass.drop('Type',axis=1),glass['Type']

le = LabelEncoder()
y = le.fit_transform(y)

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=25, stratify=y)
```

```python
from sklearn.metrics import f1_score


gbm = GradientBoostingClassifier(random_state=25)
gbm.fit(x_train,y_train)
y_pred = gbm.predict(x_test)
f1_score(y_test,y_pred, average='macro')
```

```python
trees = [50, 100, 150, 200]
rates = [0.01, 0.1, 0.2, 0.5, 0.8]
depths = [None, 2, 3, 4, 5]
scores = []
for t in tqdm(trees):
    for r in rates:
        for d in depths:
            gbm = GradientBoostingClassifier(random_state=25, learning_rate=r, max_depth=d, n_estimators=t)
            gbm.fit(x_train,y_train)
            y_pred = gbm.predict(x_test)
            scores.append([t,d, r,f1_score(y_test,y_pred, average='macro')])

df_score = pd.DataFrame(scores,columns=['trees', 'depth','learning rate','score'])
df_score.sort_values('score',ascending=False)   
```

```python
adb = AdaBoostClassifier(random_state=25)
adb.fit(x_train, y_train)
y_pred = adb.predict(x_test)
print(accuracy_score(y_test, y_pred))
```
