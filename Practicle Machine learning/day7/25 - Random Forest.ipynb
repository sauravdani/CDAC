```python
glass = pd.read_csv('C:/Users/dbda.STUDENTSDC/Desktop/M.L/Cases/Glass_Identification/Glass.csv')
glass
```

```python
x,y = glass.drop('Type',axis=1),glass['Type']
# ss = StandardScaler()
# x = ss.fit_transform(x)

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=25, stratify=y)
```

```python
features = [2,3,4,5]
n_est = [25,50,100,150,200]
scores = []

for n in n_est:
    for f in features:    
        rf = RandomForestClassifier(random_state=25,max_features=f,n_estimators=n)
        rf.fit(x_train,y_train)
        y_pred = rf.predict(x_test)
        scores.append([f,n,accuracy_score(y_test,y_pred)])
df_score = pd.DataFrame(scores,columns=['features','trees','score'])
df_score.sort_values('score',ascending=False)        
```

```python

best_tree = RandomForestClassifier(random_state=25, max_features=3, n_estimators=150)
best_tree.fit(x, y)

df_imp = pd.DataFrame({'Feature': list(x.columns), 'Importance': best_tree.feature_importances_})
df_imp = df_imp.sort_values('Importance')
plt.barh(df_imp['Feature'], df_imp['Importance'])
plt.xlabel('Importance')
plt.ylabel('Features')
plt.show()
```
